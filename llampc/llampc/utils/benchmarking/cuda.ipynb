```python
import jax

# List all available devices
devices = jax.devices()
print("Available devices:", devices)

# Check the default device
default_device = jax.default_backend()
print("Default backend:", default_device)

# Example: Verify if GPU (CUDA) is available
if any(device.platform == "gpu" for device in devices):
    print("✅ JAX is using CUDA-enabled GPU.")
```
